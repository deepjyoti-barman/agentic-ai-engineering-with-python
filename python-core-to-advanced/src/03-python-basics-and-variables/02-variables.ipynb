# Variables

A **variable** is a labelled box that holds a value. You give the box a name with `=`, and then you can use that name to read the value or change it.

```python
num1 = 8 ** 3   # creates the variable num1 and stores 512 in it
num1 + 8        # uses num1 later: 520
```

**Notes:**

- Assignment (`=`) stores a value and does not print anything. That is why the cell `num1 = 8 ** 3` has no output.
- `_` holds the last output in an interactive shell or Jupyter. In the cell `5 + _`, `_` is `512` from the cell above. This does not work in a regular `.py` script.
- Variables can be reassigned to new values, and Python decides the type from the value. Here `num3` is reassigned from `num1 + num2 + 10`.
- Names are case-sensitive (`num1` and `Num1` are different) and cannot start with a digit (`1num` is invalid).
