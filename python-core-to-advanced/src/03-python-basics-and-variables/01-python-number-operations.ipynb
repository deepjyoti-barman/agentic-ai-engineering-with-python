# Basic Number Operations

This notebook covers the arithmetic operators in Python and how they behave with integers and floats.

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `2 + 3` | `5` |
| `-` | Subtraction | `6 - 3` | `3` |
| `*` | Multiplication | `5 * 2` | `10` |
| `**` | Exponentiation (power) | `8 ** 3` | `512` |
| `/` | True division (always returns a float) | `9 / 2` | `4.5` |
| `//` | Floor division (rounds down to an integer) | `9 // 2` | `4` |
| `%` | Modulo (remainder) | `9 % 2` | `1` |
| `^` | Bitwise XOR, **not** power | `8 ^ 3` | `11` |

**Notes:**

- Use `**` for powers. `^` is a bitwise operator, so `8 ^ 3` does not mean 8 cubed.
- `/` returns a float even when the result is whole (`8 / 2` gives `4.0`).
- Multiplication and division run before addition and subtraction, and parentheses override the order. `9 + 2 * 3` is `15`, but `(9 + 2) * 3` is `33`.
- For negative numbers, `//` rounds toward negative infinity. For example, `-9 // 2` is `-5`, not `-4`.
