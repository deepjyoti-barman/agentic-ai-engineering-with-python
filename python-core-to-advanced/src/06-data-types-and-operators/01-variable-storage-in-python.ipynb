# Variable Storage in Python

A **variable** is a name that refers to an object in memory. The `=` sign does not copy a value into a box; it makes the name point to an object. Two names can point to the same object.

| Expression | Meaning |
|---|---|
| `x = 5` | Creates the name `x` and makes it refer to the object `5` |
| `id(x)` | The memory address of the object that `x` refers to |
| `x is y` | `True` if `x` and `y` refer to the same object |
| `x == y` | `True` if the two values are equal, even when they are different objects |
| `x = 6` (reassigning) | `x` now refers to the object `6`; the old object is not changed |
| `sys.intern(s)` | Returns the shared copy of the string `s` |

**Notes**

- Assigning one name to another (`k = a`) copies the reference, not the value. Both names then refer to the same object.
- `int` and `str` objects are immutable. Reassigning a name never changes the object; it only changes which object the name refers to.
- When no name refers to an object any more, Python frees its memory. CPython does this with reference counting, and a garbage collector handles cycles.
- Python shares some objects to save memory. Integers from `-5` to `256` are cached, so they are always the same object. Larger integers, and strings, are shared only when equal literals are written in the same code cell.
- Identifier-like strings, such as `"Navin"` or `"hello_world"`, are interned automatically, so equal values share one object. Strings with spaces, and strings built at runtime, are separate objects.
- Interning is an optimisation, not a guarantee. Compare values with `==`, and use `is` only to check identity, such as `x is None`.
- The `id()` numbers change every time you run the notebook, so only compare them with each other.

## String Interning

- In Python strings are stored in string pool.
- If two strings have the same value, they will refer to the same object in memory - this is called string interning.
- **String interning** is a method of storing only one copy of each distinct string value, which must be immutable.
- Identifier-like strings are interned: equal values share one object.
- Interning is an optimization, not a guarantee, so do not rely on it for comparisons.
- This can save memory and improve performance by reusing existing string objects instead of creating new ones.

**Strings Automatically Interned**
- **Empty and single-character strings.** The empty string ("") and all single-character strings in the Latin-1 range (ASCII and extended ASCII, code points 0–255) are pre-allocated as singletons and always interned.
- **Identifier-like compile-time literals.** Any string literal known at compile time that contains only ASCII letters (a-z, A-Z), digits (0-9), and underscores (_) is automatically interned.
- **Code identifiers and attribute keys.** Variable names, function names, class names, module names, and the string keys used in object attribute dictionaries are automatically interned
- **Compile-time constant-folded strings.** Expressions evaluated at compile time—such as "hello" + "_world" or "a" * 4096—are interned as long as the resulting string meets the character rules and does not exceed 4,096 characters


**Strings NOT Automatically Interned**
- Strings with spaces or punctuation.
- Strings created at runtime, such as with concatenation or formatting - that is via input(), file reading, .join(), f-strings with runtime variables, or slicing.
- Non-ASCII multi-character strings (e.g., "café" or emoji).
- Strings exceeding 4,096 characters.

## Integer Interning (Small Integer Cache)
- In Python 3.14 and earlier, integers from -5 to 256 are cached, so they are always the same object.
- In Python 3.15, CPython expanded the small integer cache from -5 to 256 to -5 to 1024. This means that integers in this range will always refer to the same object in memory, while integers outside this range may refer to different objects even if they have the same value.

```python
# a and b both refer to the object 5
a = 5
b = 5
```

```python
# Memory address of the object that a refers to
id(a)
```

```python
# Same address as a, because b also refers to the object 5
id(b)
```

```python
# b now refers to a new object 6; the object 5 is still referred to by a
b = 6

# A new address, because b refers to 6 now
id(b)
```

```python
# k refers to the existing object 5; it is not a copy
k = 5

# b now refers to the object 9
b = 9
```

```python
# Same address as a, because k refers to the object 5
id(k)
```

```python
# A different address, because b now refers to the object 9
id(b)
```

```python
# The object 5 is referred to by the variables a and k
# The object 9 is referred to by the variable b
# The object 6 is no longer referred to by any variable, so Python frees it from memory using reference counting and garbage collection
```

```python
# Identifier-like strings are interned: equal values share one object
name1 = "Navin"
name2 = "Navin"
```

```python
# Same address: both names refer to one shared string
id(name1)
```

```python
# Same address: both names refer to one shared string
id(name2)
```

```python
# Strings with spaces are not interned automatically, so equal values can be separate objects

sentence1 = "My favourite color is blue"
```

```python
# A separate cell creates a separate object for sentence2, even with the same text
sentence2 = "My favourite color is blue"
```

```python
# Address of sentence1
id(sentence1)
```

```python
# Different address: sentence2 is a separate object with the same value
id(sentence2)
```

```python
# (Python 3.15) Integers from -5 to 1024 are cached, but larger integers are not, so equal values can be separate objects
num1 = 1025
```

```python
# A separate cell creates a new object for num2, even with the same value
num2 = 1025
```

```python
# Address of num1
id(num1)
```

```python
# Different address: num2 is a separate object with the same value
id(num2)
```
