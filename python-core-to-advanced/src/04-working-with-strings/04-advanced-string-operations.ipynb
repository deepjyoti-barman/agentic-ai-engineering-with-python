# Advanced String Operations

Python strings come with many built-in **methods**. A method is called with a dot after the string, such as `text.upper()`. Every string method returns a new string (or a value); the original string is never changed, because strings are immutable.

| Method / operator | What it does | Example |
|---|---|---|
| `upper()`, `lower()`, `title()`, `capitalize()`, `swapcase()` | Change letter case | `"python".title()` gives `'Python'` |
| `strip()`, `lstrip()`, `rstrip()` | Remove whitespace (or given characters) from the ends | `"  hi  ".strip()` gives `'hi'` |
| `in`, `not in` | Check if a substring exists | `"a" in "cat"` gives `True` |
| `find()`, `rfind()` | Position of a substring, or `-1` if missing | `"banana".find("na")` gives `2` |
| `index()`, `rindex()` | Same as `find()`, but raises `ValueError` if missing | `"banana".index("na")` gives `2` |
| `count()` | Number of non-overlapping occurrences | `"banana".count("a")` gives `3` |
| `startswith()`, `endswith()` | Check the start or end of a string | `"hello".endswith("lo")` gives `True` |
| `replace(old, new)` | Replace every occurrence | `"a-b".replace("-", "+")` gives `'a+b'` |
| `split()`, `rsplit()`, `splitlines()` | Break a string into a list | `"a,b".split(",")` gives `['a', 'b']` |
| `partition()`, `rpartition()` | Split into exactly three parts around a separator | `"k=v".partition("=")` gives `('k', '=', 'v')` |
| `join()` | Join a list of strings with a separator | `"-".join(["a", "b"])` gives `'a-b'` |
| `isdigit()`, `isalpha()`, `isalnum()`, `isspace()`, `isupper()`, `islower()` | Return `True` or `False` for the content | `"123".isdigit()` gives `True` |
| `center()`, `ljust()`, `rjust()`, `zfill()` | Pad a string to a width | `"42".zfill(5)` gives `'00042'` |
| `[start:stop:step]` | Slice with a step size | `"python"[::-1]` gives `'nohtyp'` |
| f-strings, `format()` | Insert values into text, with formatting | `f"{3.14159:.2f}"` gives `'3.14'` |
| `ord()`, `chr()` | Convert a character to its code number, and back | `ord("A")` gives `65` |

**Notes**

- Methods can be chained, because each one returns a string: `"  Hello World  ".strip().lower()` gives `'hello world'`.
- `find()` returns `-1` when the substring is missing, while `index()` raises a `ValueError`. Use `find()` when missing text is a normal case.
- `split()` with no argument splits on any run of whitespace and drops empty pieces. With a separator such as `","`, empty pieces are kept.
- `isdigit()`, `isalpha()` and similar methods return `False` for an empty string.
- String comparison works character by character using code numbers, so `"10" < "9"` is `True`. Convert to `int` first when comparing numbers.
- An f-string is written with an `f` before the quotes. Its `{}` parts are replaced by values, and `=` after an expression prints both the text and the value (`{name=}`).

```python
# Changing case
text = "python is fun"
print(text.upper())       # PYTHON IS FUN
print(text.title())       # Python Is Fun
print(text.capitalize())  # Python is fun
print("PyThOn".swapcase())  # pYtHoN
```

```python
# Removing whitespace from the ends
padded = "   Hello, Python World!   "
print(repr(padded.strip()))   # Both ends
print(repr(padded.lstrip()))  # Left side only
print(repr(padded.rstrip()))  # Right side only
print("xxhixx".strip("x"))    # Removes the given characters: hi
```

```python
# Checking for a substring
course = "Python for Agentic AI"
print("Python" in course)      # True
print("Java" not in course)    # True
print("python" in course)      # False, the match is case-sensitive
```

```python
# Finding positions and counting
word = "banana"
print(word.find("na"))     # 2, first match
print(word.rfind("na"))    # 4, last match
print(word.find("z"))      # -1, not found
print(word.count("a"))     # 3
print(word.count("na"))    # 2
```

```python
# index() raises ValueError when the substring is missing
word.index("z")
```

```python
# Checking the start and end of a string
filename = "report_2026.pdf"
print(filename.endswith(".pdf"))           # True
print(filename.startswith(("report", "invoice")))  # True, tuple of options
print(filename.endswith((".doc", ".txt")))  # False
```

```python
# Replacing text
print("2026-11-28".replace("-", "/"))  # 2026/11/28
print("aaaa".replace("a", "b", 2))    # bbaa, only the first two

# replace() returns a new string; the original is unchanged
original = "cat"
changed = original.replace("c", "b")
print(original, changed)  # cat bat
```

```python
# split() and join() are opposites
sentence = "Python  is   fun"
print(sentence.split())        # ['Python', 'is', 'fun'], any whitespace
print("a,b,,c".split(","))      # ['a', 'b', '', 'c'], empty piece kept
print("key=value=extra".split("=", 1))  # ['key', 'value=extra'], maxsplit=1
print(", ".join(["Asha", "Ravi", "Mia"]))  # Asha, Ravi, Mia
```

```python
# splitlines() breaks text into lines
log = "line one\nline two\nline three"
for line in log.splitlines():
    print(line)
```

```python
# partition() splits into three parts: before, separator, after
print("username=asha".partition("="))   # ('username', '=', 'asha')
print("a=b=c".partition("="))           # ('a', '=', 'b=c'), first separator
print("a=b=c".rpartition("="))          # ('a=b', '=', 'c'), last separator
print("abc".partition("z"))             # ('abc', '', ''), separator not found
```

```python
# Checking the content of a string
print("12345".isdigit())   # True
print("abc".isalpha())    # True
print("abc123".isalnum()) # True, letters or digits only
print("   ".isspace())    # True
print("ABC".isupper())    # True
print("".isdigit())       # False, empty string
```

```python
# A common use: validating user input before converting it
age_input = "25"
if age_input.isdigit():
    print("Age is", int(age_input) + 1)
else:
    print("Please enter a number")
```

```python
# Padding and alignment (the second argument is the fill character)
print("Hi".center(10, "*"))  # ****Hi****
print("Hi".ljust(6, "."))   # Hi....
print(repr("Hi".rjust(6)))  # '    Hi'
print("42".zfill(5))        # 00042
print("-42".zfill(5))       # -0042, the sign stays first
```

```python
# Printing a neat table with padding
rows = [("Sachin", 100), ("Dravid", 50)]
for name, runs in rows:
    print(name.ljust(10) + str(runs).rjust(5))
```

```python
# Slicing with a step: start:stop:step
word = "python"
print(word[::2])    # pto, every second character
print(word[1:5:2])  # yh
print(word[::-1])   # nohtyp, reversed
print(word[::-2])   # nhy, every second character from the end
```

```python
# Looping over characters
vowels = 0
for ch in "Agentic AI".lower():
    if ch in "aeiou":
        vowels += 1
print("Vowels:", vowels)  # 5
```

```python
# Comparing strings
print("apple" < "banana")     # True, alphabetical
print("Zebra" < "apple")      # True, uppercase letters come first
print("apple" < "Zebra".lower())  # True, both sides lowercase first
print("10" < "9")             # True, compared character by character
```

```python
# f-strings: insert values and format them
name, score = "Asha", 93.456
print(f"{name} scored {score:.2f}")  # Asha scored 93.46
print(f"[{name:>8}]")                # right-aligned in 8 characters
print(f"[{name:^8}]")                # centred
print(f"{1234567:,}")                # 1,234,567
print(f"{0.25:.0%}")                 # 25%
print(f"{name=}")                    # name='Asha'
```

```python
# format() method and the older % style
print("{} is {}".format("Asha", 20))   # Asha is 20
print("{0}-{0}".format("x"))           # x-x, reuse by position
print("%s has %d runs" % ("Asha", 3))  # Asha has 3 runs
```

```python
# ord() and chr(): character codes
print(ord("A"))          # 65
print(chr(97))           # a
print(chr(ord("a") + 1))  # b, the next letter
```

```python
# Chaining methods to clean up input
raw = "  Hello   World  "
clean = raw.strip().lower()
print(repr(clean))              # 'hello   world'
print(" ".join(clean.split()))  # hello world, extra spaces removed
```

```python
# Practical example: palindrome check ignoring case and spaces
phrase = "Never odd or even"
cleaned = phrase.replace(" ", "").lower()
print(cleaned == cleaned[::-1])  # True
```
