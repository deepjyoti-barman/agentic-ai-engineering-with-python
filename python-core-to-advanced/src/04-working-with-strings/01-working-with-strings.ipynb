# Working with Strings

A **string** is text stored inside quotes. Single quotes (`'...'`) and double quotes (`"..."`) work the same way.

```python
greeting = "Hello, World!"
```

**Quotes**

- To put an apostrophe inside a string, wrap it in double quotes: `"Sachin's Book of Cricket"`. Using single quotes here would end the string early.
- Triple quotes (`'''` or `"""`) keep line breaks exactly as you type them, which makes them useful for multi-line text.

**Escape sequences** (a backslash changes the meaning of the character after it)

- `\'` inserts a single quote inside single quotes: `'Sachin\'s "Book of Cricket"'`
- `\n` is a new line, `\t` is a tab, and `\\` is one backslash.
- Windows paths need doubled backslashes: `"c:\\Users\\Sachin"`. Writing `"c:\Users"` would fail, because `\U` starts a Unicode escape. A raw string (`r"c:\Users"`) avoids escaping altogether.

**Output notes**

- A string on the last line of a cell is shown with its quotes (`'Python'`). `print()` shows the text itself.
- The `*` operator repeats a string: `"Sachin " * 10` gives the name ten times.
- Use `print()` to see how escape sequences look. A plain string expression shows them as they are written, such as `\n`.
