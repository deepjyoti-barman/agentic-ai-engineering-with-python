# Accessing Characters in a String

A string is an ordered sequence of characters. Each character has a position called an **index**, and the first index is `0`.

| Expression | Meaning | Example (`text = "telusko"`) |
|---|---|---|
| `text[i]` | Character at index `i` | `text[0]` gives `'t'` |
| `text[-i]` | Counts from the end; `-1` is the last character | `text[-1]` gives `'o'` |
| `text[start:stop]` | Slice from `start` up to, but not including, `stop` | `text[3:7]` gives `'usko'` |
| `text[:stop]` / `text[start:]` | Omit `start` to begin at `0`; omit `stop` to go to the end | `text[:3]` gives `'tel'` |
| `len(text)` | Number of characters | `len(text)` gives `7` |

**Notes**

- A valid index runs from `0` to `len(text) - 1` (here `0` to `6`), or from `-1` to `-len(text)` for negative indexes (here `-1` to `-7`).
- An index outside that range, such as `text[7]` or `text[-9]`, raises an `IndexError`.
- Slices never raise `IndexError`. A slice outside the range returns an empty string, so `text[10:]` gives `''`.
- Strings are **immutable**. `text[0] = "T"` raises a `TypeError`. To change part of a string, build a new one, as in `text[:3].upper() + text[3:]`.

```python
text = "telusko"
```

```python
text[0]  # First character
```

```python
text[3]   # Fourth character
```

```python
text[7]  # Out of range, will raise IndexError
```

```python
text[-1]  # Last character
```

```python
text[-9]  # Out of range, will raise IndexError
```

```python
text[3:7]   # Slice from index 3 to 6
```

```python
text[3:]    # Slice from index 3 to the end
```

```python
text[0:3]   # Slice from index 0 to 2
```

```python
text[:3]    # Slice from start to index 2
```

```python
text[0:3] = "TEL"  # Strings are immutable, this will raise TypeError
```

```python
text[:3].upper() + text[3:]  # Uppercase first three characters and concatenate with the rest of the string
```

```python
len(text)  # Length of the string
```
