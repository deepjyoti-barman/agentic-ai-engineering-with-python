# String Concatenation

**Concatenation** means joining two or more strings into one. Python offers three ways to do it.

**1. Adjacent string literals**

- Two string literals written next to each other are joined automatically: `"Sachin " "Tendulkar"` gives `'Sachin Tendulkar'`.
- This only works with literals written in the code, not with variables. A variable needs `+`.

**2. The `+` operator**

- `name + "Dravid"` joins a variable and a literal, giving `'Rahul Dravid'`.
- Both sides must be strings. `"Age: " + 25` raises a `TypeError`. Convert the number first with `str()`: `"Age: " + str(25)`.

**3. Missing operator (common mistake)**

- `name "Dravid"` has no `+` between a variable and a literal, so Python raises a `SyntaxError`. Adjacent joining only works between literals.

**Other notes**

- `+=` appends to an existing string: `a = "x"; a += "y"` makes `a` equal to `"xy"`.
- Each `+` creates a new string. For joining many pieces in a loop, `" ".join(list)` is more efficient.
