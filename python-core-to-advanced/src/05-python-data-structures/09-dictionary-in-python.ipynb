# Dictionary in Python

A **dictionary** stores **key-value pairs**. You look up a value by its key instead of by a position, which makes it useful for records such as name-to-score or config settings.

| Topic | Syntax | What it does |
|---|---|---|
| Create | `{key: value}` or `dict(zip(keys, values))` | Builds a dictionary |
| Access | `d[key]` | Returns the value; raises `KeyError` if the key is missing |
| Safe access | `d.get(key)`, `d.get(key, default)` | Returns `None` (or `default`) if the key is missing |
| Membership | `key in d` | Checks the keys only, not the values |
| Update / add | `d[key] = value` | Overwrites the value if the key exists, otherwise adds the pair |
| Remove | `d.pop(key)`, `del d[key]` | Removes the pair; `pop` also returns the value |
| Nested | `d[key1][key2]` | Reads a value inside another dictionary or list |

**Notes**

- Keys must be hashable (immutable), such as numbers, strings, and tuples. A list as a key raises `TypeError`.
- Values can be any type, including lists, dictionaries, and other objects.
- Keys are unique. Assigning a repeated key replaces its value and keeps the key's original position.
- `pop(key)` and `del d[key]` raise `KeyError` for a missing key. `pop(key, default)` returns `default` instead.
- Since Python 3.7, dictionaries keep insertion order. A set has no order, so `zip(set, list)` pairs items arbitrarily; use lists or tuples for keys when the pairing matters.

```python
# Creating a dictionary
data = {0: 34, 1:35, 2:67, 3: 8}
data
```

```python
# Accessing values from a dictionary with integer keys
data[2]
```

```python
# Dictionary with string keys
data = {"kiran": 34, "sushil": 35, "harsh": 67, "navin": 8}
data
```

```python
# Accessing values from a dictionary with string keys
data["harsh"]
```

```python
# Accessing values from a dictionary with string keys using get() method
data.get("harsh")
```

```python
# Follwing will throw KeyError as "praveen" is not present in the dictionary
data["praveen"]
```

```python
# The get() method returns None if the key is not found in the dictionary
data.get("praveen")
```

```python
print(data.get("praveen"))
```

```python
# get() with default value
data.get("praveen", "Not Found")
```

```python
data = {"kiran": 34, "sushil": 35, "harsh": 67, "navin": 8, "harsh": 37}
```

```python
# Dictionary keys are set, values are list
# Hence it will update the values if the keys repeat
data
```

```python
# Creating a dictionary from two lists using zip() function
keys = {"navin", "hitesh", "saumik"}
values = [34, 64, 87]
dict1 = dict(zip(keys, values))
```

```python
# When you create a dictionary from a set and a list using zip() function, order of keys and values may not be maintained in the dictionary
# If you want to maintain the order of keys and values, both keys and values should be in the form of list or tuple
print(dict1)
```

```python
# Deleting a key-value pair from a dictionary which is present using pop() method
data.pop("navin")
```

```python
data
```

```python
# Deleting a key-value pair from a dictionary which is not present using pop() method will throw KeyError
data.pop("sachin")
```

```python
# Using del keyword to delete a key-value pair from a dictionary
del data["sushil"]
```
