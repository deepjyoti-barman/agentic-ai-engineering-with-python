# Built-in Functions for Lists in Python

Python gives lists a few methods, and a few built-in functions, that work on the whole list. Some methods change the list in place, and the built-in functions only read it.

| Function / syntax | What it does | Example (`nums = [23, 55, 44, 56, 11, 99]`) |
|---|---|---|
| `list[a:b] = [...]` | Replaces items from index `a` up to `b - 1` | `nums[2:4] = [54, 76]` |
| `reverse()` | Reverses the list in place | `nums.reverse()` |
| `sort()` | Sorts the list in place, ascending | `nums.sort()` |
| `sorted(list)` | Returns a new sorted list and leaves the original alone | `sorted(nums)` |
| `min(list)` | Smallest item | `min(nums)` gives `11` |
| `max(list)` | Largest item | `max(nums)` gives `99` |
| `sum(list)` | Total of the items | `sum(nums)` gives `318` |

**Notes**

- `reverse()` and `sort()` change `nums` and return `None`. So `x = nums.sort()` gives `x` the value `None`. Use `sorted(nums)` when you want a new list.
- Slice assignment can change the length: `x = [1, 2, 3]; x[1:2] = [7, 8, 9]` gives `[1, 7, 8, 9, 3]`.
- `min()` and `max()` on strings compare them alphabetically. Uppercase letters sort before lowercase, so `min(["Navin", "harsh"])` gives `"Navin"`.
- `sum()` only adds numbers. `sum(names)` raises a `TypeError`, and so does `sum(names, "")`. To join strings, use `"".join(names)`.
- `min()` or `max()` on an empty list raises a `ValueError`.
- Comparing a string with a number, such as `sorted([1, "a"])`, raises a `TypeError`. Keep the items in a list the same type.

```python
nums = [23, 55, 44, 56, 11, 99]
names = ["navin", "harsh", "kiran"]
```

```python
nums[2:4] = [54, 76]        # Replacing elements from index 2 to 3 with new values
nums
```

```python
nums.reverse()              # Reversing the order of elements in the list
nums
```

```python
nums.sort()                 # Sorting the elements of the list in ascending order
nums
```

```python
min(nums)                   # Finding the minimum value in the list
```

```python
max(nums)                   # Finding the maximum value in the list
```

```python
sum(nums)                   # Calculating the sum of all elements in the list
```

```python
min(names)                   # Finding the minimum value in the list of strings (lexicographically)
```

```python
sum(names)
```
