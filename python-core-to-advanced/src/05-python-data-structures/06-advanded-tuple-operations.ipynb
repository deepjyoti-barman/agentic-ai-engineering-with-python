# Advanced Tuple Operations in Python

Once you know how to create and read a tuple, these operations help you slice, combine, convert, and use tuples in real programs. Tuples are immutable, so each operation below returns a new tuple or a new value.

| Topic | Syntax | What it does |
|---|---|---|
| Negative index | `tup[-1]` | Counts from the end; `-1` is the last item |
| Slice | `tup[start:stop:step]` | Returns a new tuple with the chosen items |
| Concatenate | `tup1 + tup2` | Joins two tuples into a new tuple |
| Repeat | `tup * n` | Repeats the items `n` times |
| Convert | `tuple(iterable)`, `list(tup)` | Builds a tuple, or a mutable list |
| Loop | `for`, `enumerate()`, `zip()` | Visits items with positions, or in pairs |
| Swap | `a, b = b, a` | Swaps two values without a temporary variable |
| Multiple returns | `return x, y` | A function returns a tuple that the caller unpacks |
| Sort | `sorted(tup)` | Returns a **list**; wrap it in `tuple()` to get a tuple |
| Named fields | `namedtuple` | Gives tuple items names, such as `p.x` |

**Notes**

- Slicing and `+` never change the original tuple. Print the original after each example to confirm.
- To change an item, convert the tuple to a list, change the list, then convert it back with `tuple(list)`.
- `del tup[0]` raises a `TypeError`, but `del tup` removes the name `tup` itself.
- Tuples can be dictionary keys because they are hashable. A tuple that contains a list cannot be a key.
- Tuples compare item by item from the left: `(1, 9) < (2, 0)` is `True`.
- `zip()` stops at the shortest input, so extra items in the longer tuple are dropped.

```python
# Setting up a tuple of numbers and a tuple of names
nums = (10, 20, 30, 40, 50)
names = ('Alice', 'Bob', 'Charlie')
```

```python
# Negative indexes count from the end of the tuple
nums[-1], nums[-2]
```

```python
# Slicing returns a new tuple with the items at index 1, 2 and 3
nums[1:4]
```

```python
# Step and reverse slices
nums[::2]    # Every second item
nums[::-1]   # Reversed copy
```

```python
# A slice that starts past the end gives an empty tuple, not an error
nums[10:]
```

```python
# Concatenation creates a new tuple
nums + (60, 70)
```

```python
# The original tuple is unchanged
nums
```

```python
# Repeating the items three times
(0, 1) * 3
```

```python
# Converting to a list, changing it, then converting back to a tuple
temp = list(nums)
temp[0] = 99
tuple(temp)
```

```python
# Converting other iterables into tuples
tuple([1, 2, 3]), tuple('abc'), tuple(range(3))
```

```python
# An empty tuple can be written as () or tuple()
empty = ()
type(empty), len(tuple())
```

```python
# enumerate() gives the position with each item; start=1 counts from 1
for position, name in enumerate(names, start=1):
    print(position, name)
```

```python
# zip() pairs the items of two tuples
list(zip(names, nums))
```

```python
# Swapping two values with tuple packing and unpacking
a, b = 1, 2
a, b = b, a
print(a, b)
```

```python
# A function can return several values as a tuple
def min_max(values):
    return min(values), max(values)

lowest, highest = min_max(nums)
print(lowest, highest)
```

```python
# Looping over a tuple of tuples
students = (('Alice', 90), ('Bob', 85))
for name, score in students:
    print(name, score)
```

```python
# sorted() returns a list, so wrap it in tuple() to keep a tuple
sorted_nums = tuple(sorted(nums, reverse=True))
sorted_nums
```

```python
# Tuples are compared item by item from the left
(1, 9) < (2, 0), ('apple', 2) < ('apple', 5)
```

```python
# Tuples can be dictionary keys because they are hashable
point = (3, 4)
visited = {point: 'start'}
visited
```

```python
# A list is not hashable, so it cannot be a dictionary key
try:
    {[3, 4]: 'start'}
except TypeError as error:
    print('Error:', error)
```

```python
# Deleting an item from a tuple is not allowed
try:
    del nums[0]
except TypeError as error:
    print('Error:', error)
```

```python
# namedtuple gives each position a name, so p.x works instead of p[0]
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(p, p.x, p[1])
```
