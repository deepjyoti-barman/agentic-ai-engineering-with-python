# Introduction to Tuples in Python

A tuple is an ordered collection of items written with parentheses or commas. Unlike a list, a tuple is **immutable**: once created, you cannot add, remove, or replace its items. Use a tuple for data that should stay fixed, such as a date or a coordinate.

| Function / syntax | What it does | Example (`tup = (23, 45, 67, 89)`) |
|---|---|---|
| `tup = 23, 45` | Creates a tuple without parentheses | `tup` gives `(23, 45, 67, 89)` |
| `len(tup)` | Number of items | `len(tup)` gives `4` |
| `min(tup)` / `max(tup)` | Smallest / largest item | `max(tup)` gives `89` |
| `sum(tup)` | Sum of all items | `sum(tup)` gives `224` |
| `tup.count(x)` | Number of times `x` appears | `tup.count(45)` gives `1` |
| `tup.index(x)` | Position of the first `x` | `tup.index(67)` gives `2` |
| `x in tup` / `x not in tup` | Checks whether `x` is present | `45 in tup` gives `True` |
| `a, b, c, d = tup` | Unpacks items into variables | `a` gets `23` |
| `a, *rest = tup` | Unpacks the first item and collects the rest | `rest` gets `[45, 67, 89]` |

**Notes**

- A comma creates a tuple, and parentheses only group it. `tup = (45)` is an `int`, so use `tup = (45,)` for a single-item tuple.
- Assigning to an item, such as `tup[0] = 99`, raises a `TypeError`, because tuples are immutable.
- `tup.index(x)` raises a `ValueError` when `x` is missing, so check with `in` first. `tup.count(x)` returns `0` instead.
- Unpacking needs the same number of variables as items. Otherwise Python raises a `ValueError`, such as `not enough values to unpack (expected 4, got 3)`.
- Immutability applies only to the tuple's own items. A list inside a tuple can still change, so `tupB[2][1] = 9` works.
- `min()`, `max()`, `sum()`, and `len()` return a value and never change the tuple.

```python
# Creating a tuple without parentheses
tup = 23, 45, 67, 89
type(tup)
```

```python
# Creating a tuple with parentheses
tup = (23, 45, 67, 89)
type(tup)
```

```python
# Printing the tuple
tup
```

```python
len(tup)    # Length of the tuple
```

```python
min(tup)    # Minimum value in the tuple
```

```python
max(tup)    # Maximum value in the tuple
```

```python
sum(tup)    # Sum of all values in the tuple
```

```python
tup.count(45)    # Count of a value present in the tuple
```

```python
tup.count(100)    # Count of a value not present in the tuple
```

```python
tup.index(67)    # Index of a value present in the tuple
```

```python
tup.index(100)    # Index of a value not present in the tuple
```

```python
# Tuples are immutable, so we cannot change their values
# The following code will raise a TypeError
tup[0] = 99
```

```python
tupA = (2, "naveen", 7.9)
tupA
```

```python
# Unpacking a tuple
num, name, score = tupA

print(f"num: {num}, name: {name}, score: {score}")
```

```python
# Unpacking a tuple with more than three values
# The following code will raise a ValueError because the number of variables does not match the number of values in the tuple
num, name, score1, score2 = tupA
```

```python
# Unpacking a tuple with the * operator to capture remaining values
num, *rest = tupA
print(f"num: {num}, rest: {rest}")
```

```python
tupB = (34, 'navin', [3, 4, 5, 6])
tupB[2][1] = 9
tupB
```

```python
# Checking if a value is present in the tuple
34 in tupB
```

```python
# Checking if a value is not present in the tuple
9 not in tupB
```

```python
# Creating a tuple with a single value
# But, the following code will create an integer, not a tuple
tup = (45)
type(tup)
```

```python
# To create a tuple with a single value, we need to add a comma after the value
tup = (45,)
type(tup)
```
