# Advanced List Operations in Python

Once you know the basic list methods, these operations help you search, copy, loop, and transform lists. They appear in most coding challenges.

| Topic | Syntax | What it does |
|---|---|---|
| Membership | `x in list` | `True` if `x` is in the list |
| Length | `len(list)` | Number of items |
| Copy | `list.copy()`, `list[:]`, `list(list)` | Makes a new outer list |
| Clear and delete | `list.clear()`, `del name` | Empties the list; removes the name |
| Repeat | `list * n` | Repeats the items `n` times |
| Loop | `for`, `enumerate()`, `zip()` | Visits items with positions, or in pairs |
| Step slice | `list[start:stop:step]` | Every `step`-th item; `-1` reverses |
| Sort with key | `sorted(list, key=...)` | Sorts by a rule you choose |
| Comprehension | `[expr for x in list if cond]` | Builds a new list in one line |
| Unpack | `first, *rest = list` | Splits a list into variables |
| Test | `any()`, `all()` | Checks whether any or all items are truthy |

**Notes**

- `in` checks whether a value is present and returns `True` or `False`.
- `list.copy()`, `list[:]` and `list(list)` copy only the outer list. Inner lists are shared, so use `copy.deepcopy()` for a full copy.
- `list * n` repeats references. With nested lists, `[[0] * 3] * 3` makes three references to one inner list.
- `sort()` and `sorted()` accept `key=` and `reverse=`. Python's sort is stable, so equal items keep their original order.
- An empty list is falsy, so `if not items:` is true when the list is empty.
- `map()` and `filter()` return iterators. Wrap them in `list()` to see the values.

```python
nums = [45, 87, 21, 24, 99]
names = ['Alice', 'Bob', 'Charlie', 'David']
```

```python
# Checking if a value is present
99 in nums          # True
```

```python
100 not in nums     # True
```

```python
len(names)          # Number of items in the list
```

```python
# Assignment does NOT copy: both names point to the same list
original = [1, 2, 3]
alias = original
alias.append(4)
original            # [1, 2, 3, 4]
```

```python
# Three ways to make a new (shallow) copy
copy1 = original.copy()
copy2 = original[:]
copy3 = list(original)
copy1 is original   # False: a different list object
```

```python
# A shallow copy shares the inner lists
matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
shallow[0][0] = 99
matrix              # [[99, 2], [3, 4]]
```

```python
import copy

deep = copy.deepcopy(matrix)  # Copies the inner lists too
deep[1][0] = 0
matrix              # Unchanged: [[99, 2], [3, 4]]
```

```python
# clear() empties the list, but the list still exists
temp = [1, 2, 3]
temp.clear()
temp                # []
```

```python
# del removes the name itself; using temp after this raises NameError
del temp
```

```python
zeros = [0] * 5
pattern = [1, 2] * 3
zeros, pattern     # ([0, 0, 0, 0, 0], [1, 2, 1, 2, 1, 2])
```

```python
# Trap: all three rows are the SAME inner list
grid = [[0] * 3] * 3
grid[0][0] = 1
grid                # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]
```

```python
# Correct: a new inner list for each row
good_grid = [[0] * 3 for _ in range(3)]
good_grid[0][0] = 1
good_grid           # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```

```python
# Looping over each item
for name in names:
    print(name)
```

```python
# enumerate() gives the position with each item; start=1 counts from 1
for index, name in enumerate(names, start=1):
    print(index, name)
```

```python
# zip() pairs items from two lists; it stops at the shorter list
ages = [21, 25, 30, 28]
for name, age in zip(names, ages):
    print(f'{name} is {age}')
```

```python
letters = ['a', 'b', 'c', 'd', 'e', 'f']
letters[::2]        # Every second item: ['a', 'c', 'e']
```

```python
letters[1::2]       # Every second item, starting at index 1: ['b', 'd', 'f']
```

```python
letters[::-1]       # Reversed copy: ['f', 'e', 'd', 'c', 'b', 'a']
```

```python
letters[4:1:-1]     # From index 4 down to index 2: ['e', 'd', 'c']
```

```python
words = ['banana', 'Apple', 'fig', 'cherry']
sorted(words)                  # Default: compares character codes, so 'A' comes before 'b'
```

```python
sorted(words, key=str.lower)   # Ignores case
```

```python
sorted(words, key=len)         # Shortest to longest
```

```python
# sort() changes the list in place; equal lengths keep their original order
words.sort(key=len, reverse=True)
words               # ['banana', 'cherry', 'Apple', 'fig']
```

```python
max(words, key=len), min(words, key=len)   # ('banana', 'fig'): first longest, shortest
```

```python
squares = [x * x for x in range(1, 6)]        # [1, 4, 9, 16, 25]
evens = [x for x in range(10) if x % 2 == 0]  # [0, 2, 4, 6, 8]
upper_names = [name.upper() for name in names]
labels = ['even' if x % 2 == 0 else 'odd' for x in range(4)]
labels
```

```python
# Flattening a list of lists into one list
table = [[1, 2, 3], [4, 5, 6]]
flat = [item for row in table for item in row]
flat                # [1, 2, 3, 4, 5, 6]
```

```python
list(map(str, [1, 2, 3]))                          # ['1', '2', '3']
```

```python
list(filter(lambda x: x > 50, [45, 87, 21, 99]))   # [87, 99]
```

```python
# Extra items go into a list when you use *
first, second, *rest = [10, 20, 30, 40, 50]
first, second, rest
```

```python
*head, last = [10, 20, 30]
last, head
```

```python
# Swapping two variables without a temporary variable
a, b = 1, 2
a, b = b, a
a, b                # (2, 1)
```

```python
list(range(5))          # [0, 1, 2, 3, 4]
```

```python
list(range(2, 10, 2))   # [2, 4, 6, 8]
```

```python
list('hello')           # ['h', 'e', 'l', 'l', 'o']
```

```python
# Removing duplicates while keeping the first-seen order
numbers = [3, 1, 3, 2, 1]
list(dict.fromkeys(numbers))   # [3, 1, 2]
```

```python
# any() is True if one item is truthy; all() is True only if every item is truthy
any([0, 0, 3]), all([1, 2, 0])   # (True, False): one truthy item is enough for any(); 0 is falsy
```

```python
all([])             # True: there are no items to fail the check
```

```python
any([])             # False: there are no items to pass the check
```
