# List Functions in Python

A list has built-in methods that change it in place, plus the `+` operator, which builds a new list. Methods such as `append()`, `insert()`, `remove()`, and `pop()` modify the original list and return `None` or the removed item.

| Function / syntax | What it does | Example (`nums = [45, 87, 21, 24, 99]`) |
|---|---|---|
| `list1 + list2` | Joins two lists into a new list | `nums + ["Alice"]` |
| `append(x)` | Adds `x` to the end as one item | `nums.append(33)` |
| `extend(iterable)` | Adds each item of the iterable to the end | `nums.extend([44, 56])` |
| `count(x)` | Number of times `x` appears | `nums.count(99)` gives `1` |
| `insert(i, x)` | Inserts `x` at index `i` | `nums.insert(1, 55)` |
| `remove(x)` | Removes the first item equal to `x` | `nums.remove(21)` |
| `pop()` / `pop(i)` | Removes and returns the last item, or the item at index `i` | `nums.pop(3)` |
| `del nums[a:b]` | Deletes items from index `a` up to `b - 1` | `del nums[1:3]` |

**Notes**

- `append(x)` adds `x` as a single item, so `nums.append([1, 2])` adds one nested list. `extend([1, 2])` adds the two numbers separately.
- `remove(x)` raises a `ValueError` if `x` is not in the list. `count(x)` returns `0` in that case instead of an error.
- `pop()` raises an `IndexError` on an empty list or an index that is out of range.
- `insert(i, x)` with an `i` past the end adds the item at the end; it does not raise an error.
- `nums + names` creates a new list and leaves `nums` unchanged. `append`, `extend`, `insert`, and `remove` change `nums` itself.
- `index(x)` returns the position of the **first** match, so `[3, 1, 3].index(3)` gives `0`. It raises a `ValueError` when `x` is missing, so check with `in` first if the value may be absent.
- `pop()` returns the removed item, so `last = nums.pop()` saves it. `remove()`, `insert()`, `append()`, and `extend()` return `None`.
- `del nums[1:3]` removes the items at positions 1 and 2 and returns nothing, unlike `pop()`.

```python
nums = [45, 87, 21, 24, 99]
names = ["Alice", "Bob", "Charlie", "David"]
```

```python
join = nums + names # Joining two lists
join
```

```python
nums.append(33)     # Adding an element to the end of the list
nums
```

```python
nums.count(99)      # Counting occurrences of an element which is present in the list
```

```python
nums.count(15)      # Counting occurrences of an element which is not present in the list
```

```python
nums.index(21)       # Finding the index of an element which is present in the list
```

```python
nums.index(12)       # Finding the index of an element which is not present in the list, it will raise a ValueError
```

```python
nums.insert(1, 55)   # Inserting an element at a specific index
nums
```

```python
nums.insert(11, 88)  # Inserting an element at a specific index beyond the length of the list, it will be added at the end
nums
```

```python
nums.remove(21)     # Removing an element which is present in the list
nums
```

```python
nums.remove(102)   # Trying to remove an element which is not present in the list, it will raise a ValueError
```

```python
nums.pop()          # Removing the last element from the list
nums
```

```python
nums.pop(3)         # Removing an element at a specific index which is present in the list
nums
```

```python
nums.pop(10)        # Trying to remove an element at a specific index which is not present in the list, it will raise an IndexError
```

```python
del nums[1:3]       # Deleting elements from index 1 to 2
nums
```

```python
nums.extend([44, 56, 11, 9])  # Extending the list with another list
nums
```
