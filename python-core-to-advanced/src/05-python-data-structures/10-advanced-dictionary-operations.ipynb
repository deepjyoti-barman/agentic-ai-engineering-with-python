# Advanced Dictionary Operations in Python

Once you know how to create a dictionary and read, add, or remove its key-value pairs, these operations help you loop over dictionaries, merge them, count things, and build new dictionaries in one line.

| Topic | Syntax | What it does |
|---|---|---|
| Views | `d.keys()`, `d.values()`, `d.items()` | Live views of the keys, values, or `(key, value)` pairs |
| Loop | `for k, v in d.items()` | Visits each key and value together |
| Default value | `d.setdefault(key, default)` | Returns the value; adds `key: default` first if the key is missing |
| Update | `d.update(other)`, `d \| other` | Merges `other` into `d`; values from `other` win |
| In-place merge | `d \|= other` | Same as `update()` |
| Copy | `d.copy()` | Creates an independent dictionary |
| Pop last item | `d.popitem()` | Removes and returns the last-inserted pair |
| Clear | `d.clear()` | Removes every pair |
| From keys | `dict.fromkeys(keys, value)` | Builds a dictionary with the same value for every key |
| Comprehension | `{k: v for k, v in d.items() if ...}` | Builds a new dictionary in one line |
| Unpacking | `{**d1, **d2}` | Merges dictionaries into a new one |
| Counting | `d[key] = d.get(key, 0) + 1` | Counts items without a `KeyError` |

**Notes**

- `keys()`, `values()`, and `items()` return views, not lists. A view updates when the dictionary changes. Wrap it in `list()` to get a copy.
- `update()`, `|`, and `|=` overwrite values for keys that already exist, so the value from the other dictionary wins.
- `setdefault()` only adds the key when it is missing. `get()` never adds anything.
- `popitem()` removes the last-inserted pair, not a random one.
- Do not add or remove keys while looping over the dictionary. Changing its size raises `RuntimeError`. Loop over `list(d)` instead if you must change it.
- `dict()` with keyword arguments, such as `dict(a=1, b=2)`, creates string keys without quotes.
- Dictionaries compare equal when they have the same keys and values, regardless of order.
- For counting, `collections.Counter` and grouping with `collections.defaultdict` do the same job in fewer lines.

```python
# Setting up a dictionary of student scores
scores = {'kiran': 34, 'sushil': 35, 'harsh': 67}

# keys(), values(), and items() return views of the dictionary
print(scores.keys())
print(scores.values())
print(scores.items())
```

```python
# Wrap a view in list() to see its items as a normal list
print(list(scores.keys()))

# Checking a key with in looks at the keys, not the values
print('harsh' in scores, 67 in scores, 67 in scores.values())
```

```python
# Looping over key-value pairs with items()
for name, score in scores.items():
    print(name, '->', score)
```

```python
# Looping over a dictionary directly visits its keys
for name in scores:
    print(name)
```

```python
# setdefault() returns the value and adds the key only if it is missing
print(scores.setdefault('praveen', 0))
print(scores.setdefault('kiran', 99))
print(scores)
```

```python
# copy() makes a separate dictionary, so changing the copy leaves the original alone
backup = scores.copy()
backup['sachin'] = 50
print('original:', scores)
print('copy:', backup)
```

```python
# update() adds new keys and overwrites the values of existing keys
scores.update({'harsh': 80, 'navin': 8})
scores
```

```python
# The | operator returns a new merged dictionary; the right side wins on duplicate keys
a = {'x': 1, 'y': 2}
b = {'y': 20, 'z': 3}
print(a | b)
print(b | a)
print('a is unchanged:', a)
```

```python
# The |= operator merges b into a in place
a |= b
a
```

```python
# {**d1, **d2} also merges into a new dictionary, the same as d1 | d2
print({**{'a': 1}, **{'a': 2, 'b': 3}})
```

```python
# popitem() removes and returns the last-inserted pair
last = a.popitem()
print(last)
print(a)
```

```python
# dict.fromkeys() builds a dictionary with the same starting value for each key
print(dict.fromkeys(['math', 'science'], 0))
```

```python
# A dictionary comprehension builds a new dictionary in one line
doubled = {name: score * 2 for name, score in scores.items() if score > 30}
doubled
```

```python
# Sorting keys by their values, highest first
ranking = sorted(scores, key=lambda name: scores[name], reverse=True)
print(ranking)
```

```python
# Counting words with get(); a missing word starts at 0
words = 'apple banana apple cherry apple'.split()
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

```python
# Grouping words by their first letter with setdefault()
groups = {}
for word in words:
    groups.setdefault(word[0], []).append(word)
print(groups)
```

```python
# Counter from the collections module does the same counting in one line
from collections import Counter

print(Counter(words))
print(Counter(words).most_common(1))
```

```python
# Changing the size of a dictionary while looping over it raises a RuntimeError
data = {'a': 1, 'b': 2}
try:
    for key in data:
        del data[key]
except RuntimeError as error:
    print('RuntimeError:', error)
```

```python
# Looping over a copy of the keys is safe, so the deletion works
data = {'a': 1, 'b': 2, 'c': 3}
for key in list(data):
    if data[key] % 2 == 1:
        del data[key]
print(data)
```

```python
# Dictionaries with the same keys and values are equal, even in a different order
print({'a': 1, 'b': 2} == {'b': 2, 'a': 1})
print({'a': 1} == {'a': 1, 'b': 2})

# dict() with keyword arguments creates string keys without quotes
print(dict(a=1, b=2))
```

```python
# Nested dictionaries: a dictionary whose values are dictionaries
students = {
    'kiran': {'age': 20, 'city': 'Pune'},
    'harsh': {'age': 22, 'city': 'Delhi'},
}

# Looping over the outer and inner dictionary together
for name, info in students.items():
    for field, value in info.items():
        print(name, field, value)
```
