# Set in Python

A **set** is a collection of **unique**, **unordered** items. Use it to remove duplicates or to check quickly whether an item exists.

| Topic | Syntax | What it does |
|---|---|---|
| Create | `{1, 2, 3}` or `set(iterable)` | Builds a set; duplicates are dropped |
| Empty set | `set()` | `{}` creates an empty **dict**, not a set |
| Length | `len(s)` | Counts the items |
| Membership | `x in s` | Returns `True` or `False` |
| Add / remove | `s.add(x)`, `s.remove(x)`, `s.discard(x)` | `remove` raises `KeyError` if `x` is missing; `discard` does not |
| Difference | `s1 - s2` | Items in `s1` but not in `s2` |
| Union | `s1 \| s2` | Items in either set |
| Intersection | `s1 & s2` | Items in both sets |
| Symmetric difference | `s1 ^ s2` | Items in exactly one of the sets |

**Notes**

- The set itself is mutable, so you can add and remove items. Each item inside it must be immutable: numbers, strings, and tuples work, but lists, dicts, and sets raise `TypeError`.
- Sets are unordered, so the printed order can differ from the order you typed. Indexing such as `set1[0]` raises `TypeError`.
- Python uses each item's hash value to decide where it is stored, which makes `x in s` fast even for large sets.
- `set('abc')` splits a string into its characters: `{'a', 'b', 'c'}`.

```python
set1 = {23, 56, 78, 21, 56}
```

```python
set1
```

```python
print(set1)
```

```python
# Checking the length of the set
len(set1)
```

```python
# Checking the type of the set
type(set1)
```

```python
# We cannot create an empty set using curly braces, as it will create an empty dictionary instead
set2 = {}
type(set2)
```

```python
# Creating an empty set using the set() constructor
set3 = set()
type(set3)
```

```python
set2 = set("abcdmnop")
set3 = set("aeioupd")
```

```python
print(set2)
```

```python
print(set3)
```

```python
# Performing set difference operation
set2 - set3
```

```python
# Performing set union operation
set2 | set3
```

```python
# Performing set intersection operation
set2 & set3
```

```python
# Performing set symmetric difference operation
set2 ^ set3
```
