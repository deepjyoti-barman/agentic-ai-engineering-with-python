# Advanced Set Operations in Python

Once you know how to create a set and use the union, intersection, difference, and symmetric difference operators, these operations help you change sets, compare them, and solve everyday problems such as removing duplicates.

| Topic | Syntax | What it does |
|---|---|---|
| Add | `s.add(x)` | Adds one item; adding an existing item does nothing |
| Remove | `s.remove(x)` | Removes `x`; raises `KeyError` if `x` is missing |
| Discard | `s.discard(x)` | Removes `x`; does nothing if `x` is missing |
| Pop | `s.pop()` | Removes and returns an arbitrary item; raises `KeyError` if the set is empty |
| Clear | `s.clear()` | Removes every item |
| Method forms | `s.union(x)`, `intersection`, `difference`, `symmetric_difference` | Accept any iterable, such as a list |
| In-place operators | `\|=`, `&=`, `-=`, `^=` | Change the original set |
| Update methods | `update()`, `intersection_update()`, `difference_update()`, `symmetric_difference_update()` | Change the original set using any iterable |
| Subset / superset | `issubset()` / `<=`, `issuperset()` / `>=`, `<` / `>` | Compare two sets; `<` and `>` require them to be different |
| Disjoint | `isdisjoint()` | `True` if the two sets share no items |
| Copy | `s.copy()` | Creates an independent set |
| Frozen set | `frozenset(iterable)` | An immutable set |

**Notes**

- `add`, `remove`, `discard`, `pop`, `clear`, and the in-place operators change the set itself. They return `None` (or the removed item for `pop`), not a new set.
- Use `discard()` instead of `remove()` when the item might be missing, so no `KeyError` is raised.
- The operators `|`, `&`, `-`, and `^` need two sets, but the methods such as `union()` also accept lists and other iterables.
- Sets have no order, so `pop()` removes an arbitrary item. Use `sorted()` when you need the items in order.
- String sets can print in a different order on each run, so the examples below use `sorted()` when they print strings.
- Assignment shares one set between two names, while `copy()` makes a separate set.
- `set(items)` removes duplicates but loses the original order. `dict.fromkeys(items)` removes duplicates and keeps the first-seen order.

```python
# Setting up a set of numbers
a = {1, 2, 3, 4}

# add() puts one item in the set; adding an existing item does nothing
a.add(10)
a.add(2)
a
```

```python
# remove() deletes the item, but raises a KeyError if the item is missing
a.remove(10)
try:
    a.remove(99)
except KeyError as error:
    print('KeyError:', error)
```

```python
# discard() deletes the item and does nothing if the item is missing
a.discard(99)
a
```

```python
# pop() removes and returns an arbitrary item, because sets have no order
item = a.pop()
print(item, a)
```

```python
# clear() removes every item and leaves an empty set
c = {7, 8}
c.clear()
c
```

```python
# Membership tests with in and not in
print(3 in a, 99 not in a)
```

```python
# The method forms accept any iterable, such as a list
print(a.union([7]))
print(a.intersection([3, 9]))
print(a.difference([2]))
print(a.symmetric_difference([4, 5]))
```

```python
# The | operator needs two sets, so a list raises a TypeError
try:
    print({1, 2} | [3])
except TypeError as error:
    print('Error:', error)
```

```python
# The |= operator changes x itself, the same as x = x | {4, 5}
x = {1, 2, 3}
x |= {4, 5}
x
```

```python
# update() adds every item from any iterable, the same as |=
x.update([6, 7])
x
```

```python
# intersection_update() keeps only the items that are also in the other set
x.intersection_update({1, 2, 6, 7})
x
```

```python
# difference_update() removes the items that are also in the other set
x.difference_update({2})
x
```

```python
small = {1, 2}
big = {1, 2, 3, 4}

# issubset() and <= check that every item of small is in big
print(small.issubset(big), small <= big)

# issuperset() and >= check the reverse
print(big.issuperset(small), big >= small)

# A proper subset is a subset that is not equal, so < is False for equal sets
print(small < small, small <= small)
```

```python
# isdisjoint() is True when the two sets have no items in common
print({1, 2}.isdisjoint({3, 4}))
print({1, 2}.isdisjoint({2, 3}))
```

```python
original = {1, 2, 3}

# Assignment does not copy the set, so alias and original are the same set
alias = original

# copy() makes a separate set
duplicate = original.copy()

original.add(99)
print('alias:', alias)
print('copy:', duplicate)
```

```python
fs = frozenset([1, 2, 3])
print(fs)

# A frozenset cannot change, so it can be used as a dictionary key
print({fs: 'valid key'})

# Trying to add an item raises an AttributeError
try:
    fs.add(4)
except AttributeError as error:
    print('Error:', error)
```

```python
# A set comprehension builds a set in one line, and duplicates are removed
squares = {n * n for n in range(-3, 4)}
squares
```

```python
nums = {4, 9, 1, 7}

# len(), min(), max(), and sum() work on sets
print(len(nums), min(nums), max(nums), sum(nums))

# sorted() returns a list in order, because sets have no order
print(sorted(nums))

# any() and all() check a condition for the items
print(any(n > 8 for n in nums), all(n > 0 for n in nums))
```

```python
nums_list = [3, 1, 3, 2, 1]

# set() removes duplicates, but the original order is lost
print(list(set(nums_list)))

# dict.fromkeys() removes duplicates and keeps the first-seen order
print(list(dict.fromkeys(nums_list)))
```

```python
ordered = ['apple', 'banana', 'mango']
delivered = ['banana', 'apple']

# Items in the order that were not delivered
missing = set(ordered) - set(delivered)
print(sorted(missing))

# Items that were both ordered and delivered
print(sorted(set(ordered) & set(delivered)))
```
