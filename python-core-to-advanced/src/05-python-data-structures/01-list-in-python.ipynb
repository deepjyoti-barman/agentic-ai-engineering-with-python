# List in Python

A **list** is an ordered, mutable collection of items written in square brackets `[]` and separated by commas.

## Key Notes

- **Ordered:** Items keep their insertion order, so each item has a fixed position (index).
- **Zero-based indexing:** The first item is at index `0`. Negative indexes count from the end, so `nums[-1]` is the last item.
- **Slicing:** `list[start:stop]` returns a *new* list from `start` up to, but not including, `stop`. Omitting `stop` goes to the end, so `nums[2:]` returns everything from index 2 onward.
- **Mixed types:** A list can hold values of different types, such as strings, integers, floats, and booleans.
- **Nested lists:** `[nums, names]` creates a list whose items are the two lists; it does not flatten them. Chain indexes to reach inner items, e.g. `combined[1][2]` is `'Charlie'`.
- **References, not copies:** A nested list stores references to the original lists, so changes to `nums` are visible through `combined[0]`.
- **Concatenation:** `nums + names` creates a new flat list and leaves both originals unchanged.
- **Mutable:** Lists can be changed in place (e.g. `nums[0] = 1` or `nums.append(7)`).

```python
# List is a collection of items in a particular order
nums = [45, 87, 21, 24, 99]
```

```python
# Accessing elements in a list
print(nums[0])  # First element
```

```python
print(nums[-1])  # Last element
```

```python
# Slicing a list
nums[2:4]       # Elements from index 2 to 3
```

```python
nums[2:]        # Elements from index 2 to the end
```

```python
# Creating a list of names
names = ["Alice", "Bob", "Charlie", "David"]
names
```

```python
# Creating a mixed list
mix = ["Hello", 42, 3.14, True]
mix
```

```python
# Combining two lists
combined = [nums, names]
combined
```

```python
combined[0]     # Accessing the first list (nums)
```

```python
combined[1][2]  # Accessing "Charlie"
```

```python
# Joining two lists
join = nums + names  # Joining two lists
join
```
